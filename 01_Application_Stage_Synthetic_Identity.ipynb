```python
import pandas as pd
import numpy as np
from faker import Faker
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report, recall_score
from sklearn.ensemble import IsolationForest
from sklearn.linear_model import LogisticRegression
import xgboost as xgb
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
N_SAMPLES = 20000
FRAUD_RATIO = 0.18
RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
fake = Faker('en_IN')
Faker.seed(RANDOM_STATE)

def generate_harder_data(n_samples=20000, fraud_ratio=0.18):
    data = []
    shared_devices = [fake.uuid4() for _ in range(150)]
    shared_ips = [fake.ipv4() for _ in range(110)]
    shared_phones = [fake.phone_number() for _ in range(130)]

    for i in range(n_samples):
        is_fraud = 1 if np.random.rand() < fraud_ratio else 0
        age = np.random.randint(20, 55)
        education = np.random.choice(['High School', 'Graduate', 'Post Graduate'], p=[0.3, 0.5, 0.2])

        if is_fraud:
            difficulty = np.random.choice(['easy', 'medium', 'hard'], p=[0.30, 0.25, 0.45])

            if difficulty == 'easy':
                work_exp = max(0, age - np.random.randint(7, 14))
                job = np.random.choice(['Student', 'Unemployed'])
                income = np.random.randint(12, 35) * 100000
                device = np.random.choice(shared_devices)
                ip = np.random.choice(shared_ips)
                phone = np.random.choice(shared_phones)
                address_changes = np.random.randint(3, 8)
                num_apps = np.random.randint(4, 8)
            elif difficulty == 'medium':
                work_exp = max(0, age - np.random.randint(15, 21))
                job = np.random.choice(['Salaried', 'Self-employed', 'Student'])
                income = np.random.randint(8, 25) * 100000
                device = np.random.choice(shared_devices) if np.random.rand() < 0.55 else fake.uuid4()
                ip = np.random.choice(shared_ips) if np.random.rand() < 0.4 else fake.ipv4()
                phone = fake.phone_number()
                address_changes = np.random.randint(2, 5)
                num_apps = np.random.randint(2, 5)
            else:  # hard
                work_exp = max(0, age - np.random.randint(18, 24))
                job = np.random.choice(['Salaried', 'Self-employed'])
                income = np.random.randint(5, 18) * 100000
                device = np.random.choice(shared_devices) if np.random.rand() < 0.25 else fake.uuid4()
                ip = fake.ipv4()
                phone = fake.phone_number()
                address_changes = np.random.randint(0, 3)
                num_apps = np.random.randint(1, 3)
            fraud_difficulty = difficulty
        else:
            work_exp = max(0, age - np.random.randint(19, 27))
            if np.random.rand() < 0.08:
                work_exp = max(0, work_exp - np.random.randint(1, 3))
            job = np.random.choice(['Salaried', 'Self-employed', 'Government', 'Student'], p=[0.55, 0.25, 0.1, 0.1])
            income = np.random.randint(2, 22) * 100000
            if job == 'Student':
                income = np.random.randint(0, 5) * 100000
            device = fake.uuid4()
            ip = fake.ipv4()
            phone = fake.phone_number()
            address_changes = np.random.randint(0, 3)
            num_apps = np.random.randint(1, 3)
            fraud_difficulty = 'Genuine'

        data.append({
            'applicant_id': f'APP{100000 + i}',
            'age': age,
            'work_experience': work_exp,
            'declared_income': income,
            'job_category': job,
            'education': education,
            'device_id': device,
            'ip_address': ip,
            'phone_number': phone,
            'address_change_count': address_changes,
            'num_applications': num_apps,
            'fraud_difficulty': fraud_difficulty,
            'label': is_fraud
        })

    return pd.DataFrame(data)

df = generate_harder_data(N_SAMPLES, FRAUD_RATIO)
print("Data shape:", df.shape)
print(df['label'].value_counts(normalize=True))
print("\nFraud Difficulty:")
print(df['fraud_difficulty'].value_counts())
df.head()
```

```python
# ===============================================================
#  Rules + Features + Identity Coherence Score
# ===============================================================

def apply_consistency_rules(df):
    df = df.copy()

    df['rule_age_exp'] = ((df['work_experience'] > (df['age'] - 17)) |
                          ((df['age'] <= 23) & (df['work_experience'] >= 5))).astype(int)

    df['rule_income_job'] = (((df['job_category'].isin(['Student', 'Unemployed'])) &
                              (df['declared_income'] > 750000))).astype(int)

    df['rule_velocity'] = (df['num_applications'] >= 4).astype(int)
    df['rule_address'] = (df['address_change_count'] >= 4).astype(int)

    df['rule_score'] = df[['rule_age_exp', 'rule_income_job', 'rule_velocity', 'rule_address']].sum(axis=1)
    return df

df = apply_consistency_rules(df)
print("Rules applied successfully!")
print(df[['rule_age_exp', 'rule_income_job', 'rule_velocity', 'rule_address', 'rule_score']].head())


# ---------------------------------------------------------------
# Feature Engineering (Basic version - leakage free later in CV)
# ---------------------------------------------------------------
def create_basic_features(df):
    df = df.copy()

    df['device_share'] = df.groupby('device_id')['applicant_id'].transform('count')
    df['ip_share'] = df.groupby('ip_address')['applicant_id'].transform('count')
    df['phone_share'] = df.groupby('phone_number')['applicant_id'].transform('count')

    df['device_share_log'] = np.log1p(df['device_share'])
    df['ip_share_log'] = np.log1p(df['ip_share'])
    df['phone_share_log'] = np.log1p(df['phone_share'])

    df['age_exp_gap'] = df['age'] - df['work_experience']
    df['income_per_exp'] = df['declared_income'] / (df['work_experience'] + 1)
    df['income_per_age'] = df['declared_income'] / df['age']

    return df

df = create_basic_features(df)
print("\nBasic features created!")


# ---------------------------------------------------------------
# Identity Coherence Score (ICS)
# ---------------------------------------------------------------
def calculate_identity_coherence_score(df):
    df = df.copy()

    # 1. Demographic Coherence
    demo_penalty = (
        df['rule_age_exp'] * 0.40 +
        df['rule_income_job'] * 0.35 +
        (df['age_exp_gap'] < 10).astype(int) * 0.15 +
        (df['income_per_exp'] > df['income_per_exp'].quantile(0.95)).astype(int) * 0.10
    )
    demo_penalty = np.clip(demo_penalty, 0, 1)
    df['demographic_coherence'] = 1 - demo_penalty

    # 2. Relational Coherence
    rel_penalty = (
        0.50 * df['device_share_log'] +
        0.30 * df['ip_share_log'] +
        0.20 * df['phone_share_log']
    )
    scaler = MinMaxScaler()
    rel_penalty_norm = scaler.fit_transform(rel_penalty.values.reshape(-1, 1)).flatten()
    df['relational_coherence'] = 1 - rel_penalty_norm

    # 3. Behavioral Coherence
    behav_penalty = (
        0.60 * (df['num_applications'] >= 4).astype(int) +
        0.40 * (df['address_change_count'] >= 3).astype(int)
    )
    behav_penalty = np.clip(behav_penalty, 0, 1)
    df['behavioral_coherence'] = 1 - behav_penalty

    # Final ICS
    w1, w2, w3 = 0.40, 0.35, 0.25
    df['identity_coherence_score'] = (
        w1 * df['demographic_coherence'] +
        w2 * df['relational_coherence'] +
        w3 * df['behavioral_coherence']
    )

    # Risk version
    df['identity_risk_from_ics'] = 1 - df['identity_coherence_score']

    return df

df = calculate_identity_coherence_score(df)

print("\nIdentity Coherence Score calculated!")
print(df[['applicant_id', 'demographic_coherence', 'relational_coherence',
          'behavioral_coherence', 'identity_coherence_score', 'identity_risk_from_ics']].head(10))

print("\nICS Statistics:")
print(df['identity_coherence_score'].describe())
```

```python
# ===============================================================
#  Model Training + Evaluation
# ===============================================================

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report, recall_score

# Encode categorical columns
le_job = LabelEncoder()
le_edu = LabelEncoder()
df['job_enc'] = le_job.fit_transform(df['job_category'])
df['edu_enc'] = le_edu.fit_transform(df['education'])

FEATURE_COLS = [
    'age', 'work_experience', 'declared_income', 'job_enc', 'edu_enc',
    'address_change_count', 'num_applications',
    'device_share_log', 'ip_share_log', 'phone_share_log',
    'age_exp_gap', 'income_per_exp', 'income_per_age', 'rule_score',
    'identity_coherence_score'          # <-- ICS added as feature
]

X = df[FEATURE_COLS]
y = df['label']

print("Features used:", FEATURE_COLS)
print("X shape:", X.shape)

# ---------------------------------------------------------------
# Stratified 5-Fold Cross Validation
# ---------------------------------------------------------------
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

oof_xgb = np.zeros(len(df))
oof_lgb = np.zeros(len(df))

fold = 1
for train_idx, val_idx in skf.split(X, y):
    print(f"Training Fold {fold}...")

    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    # XGBoost
    xgb_model = xgb.XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum(),
        random_state=42,
        eval_metric='auc',
        n_jobs=-1
    )
    xgb_model.fit(X_train, y_train)
    oof_xgb[val_idx] = xgb_model.predict_proba(X_val)[:, 1]

    # LightGBM
    lgb_model = lgb.LGBMClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum(),
        random_state=42,
        n_jobs=-1
    )
    lgb_model.fit(X_train, y_train)
    oof_lgb[val_idx] = lgb_model.predict_proba(X_val)[:, 1]

    fold += 1

df['xgb_proba'] = oof_xgb
df['lgb_proba'] = oof_lgb

# ---------------------------------------------------------------
# Results
# ---------------------------------------------------------------
print("\n========== OVERALL RESULTS ==========")
print("XGBoost ROC-AUC :", round(roc_auc_score(y, oof_xgb), 4))
print("XGBoost PR-AUC  :", round(average_precision_score(y, oof_xgb), 4))
print("LightGBM ROC-AUC:", round(roc_auc_score(y, oof_lgb), 4))
print("LightGBM PR-AUC :", round(average_precision_score(y, oof_lgb), 4))

print("\n========== Classification Report (XGBoost) ==========")
print(classification_report(y, (oof_xgb > 0.5).astype(int)))

# ---------------------------------------------------------------
# Performance on Easy vs Hard Fraud
# ---------------------------------------------------------------
print("\n========== RECALL BY FRAUD DIFFICULTY ==========")
for diff in ['easy', 'medium', 'hard']:
    mask = (df['fraud_difficulty'] == diff)
    if mask.sum() > 0:
        rec = recall_score(df.loc[mask, 'label'], (df.loc[mask, 'xgb_proba'] > 0.5).astype(int))
        print(f"{diff.upper():<8} Fraud Recall: {rec:.4f}   (Samples: {mask.sum()})")

print("\nStep 5 Completed Successfully!")
```

```python
# ===============================================================
# LEAKAGE-AWARE HOLDOUT EVALUATION (NEW; NOT YET EXECUTED)
# Historical CV metrics and xgb_identity_model.json remain unchanged.
# Run only after approving model training in this environment.
# ===============================================================

import numpy as np
import pandas as pd
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    average_precision_score,
    confusion_matrix,
    f1_score,
    precision_recall_curve,
    precision_score,
    recall_score,
    roc_auc_score,
)

EVAL_SEED = 42

# Start from raw generated applicant fields; ignore globally engineered columns
# from the historical experiment above.
raw_columns = [
    'applicant_id', 'age', 'work_experience', 'declared_income',
    'job_category', 'education', 'device_id', 'ip_address', 'phone_number',
    'address_change_count', 'num_applications', 'label'
]
raw_applicants = df[raw_columns].copy().reset_index(drop=True)

train_validation_raw, test_raw = train_test_split(
    raw_applicants,
    test_size=0.20,
    random_state=EVAL_SEED,
    stratify=raw_applicants['label'],
)
train_raw, validation_raw = train_test_split(
    train_validation_raw,
    test_size=0.20,
    random_state=EVAL_SEED,
    stratify=train_validation_raw['label'],
)
train_raw = train_raw.reset_index(drop=True)
validation_raw = validation_raw.reset_index(drop=True)
test_raw = test_raw.reset_index(drop=True)


def fit_feature_reference(training_rows):
    reference = {}
    for key in ('device_id', 'ip_address', 'phone_number'):
        reference[f'{key}_counts'] = training_rows.groupby(key).size().to_dict()
    reference['income_per_exp_95th'] = (
        training_rows['declared_income'] / (training_rows['work_experience'] + 1)
    ).quantile(0.95)
    return reference


def transform_applicants(rows, reference, training_rows=False):
    result = rows.copy()
    result['rule_age_exp'] = (
        (result['work_experience'] > (result['age'] - 17)) |
        ((result['age'] <= 23) & (result['work_experience'] >= 5))
    ).astype(int)
    result['rule_income_job'] = (
        result['job_category'].isin(['Student', 'Unemployed']) &
        (result['declared_income'] > 750000)
    ).astype(int)
    result['rule_velocity'] = (result['num_applications'] >= 4).astype(int)
    result['rule_address'] = (result['address_change_count'] >= 4).astype(int)
    result['rule_score'] = result[
        ['rule_age_exp', 'rule_income_job', 'rule_velocity', 'rule_address']
    ].sum(axis=1)

    for key, prefix in (
        ('device_id', 'device'),
        ('ip_address', 'ip'),
        ('phone_number', 'phone'),
    ):
        counts = reference[f'{key}_counts']
        if training_rows:
            share = result[key].map(counts).fillna(1)
        else:
            # Include the applicant itself plus matches already seen in training.
            share = result[key].map(counts).fillna(0) + 1
        result[f'{prefix}_share_log'] = np.log1p(share)

    result['age_exp_gap'] = result['age'] - result['work_experience']
    result['income_per_exp'] = result['declared_income'] / (result['work_experience'] + 1)
    result['income_per_age'] = result['declared_income'] / result['age']

    demo_penalty = (
        result['rule_age_exp'] * 0.40 +
        result['rule_income_job'] * 0.35 +
        (result['age_exp_gap'] < 10).astype(int) * 0.15 +
        (result['income_per_exp'] > reference['income_per_exp_95th']).astype(int) * 0.10
    )
    result['demographic_coherence'] = 1 - np.clip(demo_penalty, 0, 1)

    rel_penalty = (
        0.50 * result['device_share_log'] +
        0.30 * result['ip_share_log'] +
        0.20 * result['phone_share_log']
    )
    if training_rows:
        reference['rel_penalty_min'] = float(rel_penalty.min())
        reference['rel_penalty_max'] = float(rel_penalty.max())
    rel_range = reference['rel_penalty_max'] - reference['rel_penalty_min']
    if rel_range == 0:
        rel_penalty_norm = np.zeros(len(result), dtype=float)
    else:
        rel_penalty_norm = np.clip(
            (rel_penalty.to_numpy() - reference['rel_penalty_min']) / rel_range,
            0,
            1,
        )
    result['relational_coherence'] = 1 - rel_penalty_norm

    behavior_penalty = (
        0.60 * (result['num_applications'] >= 4).astype(int) +
        0.40 * (result['address_change_count'] >= 3).astype(int)
    )
    result['behavioral_coherence'] = 1 - np.clip(behavior_penalty, 0, 1)
    result['identity_coherence_score'] = (
        0.40 * result['demographic_coherence'] +
        0.35 * result['relational_coherence'] +
        0.25 * result['behavioral_coherence']
    )

    if training_rows:
        reference['job_categories'] = sorted(result['job_category'].unique())
        reference['education_categories'] = sorted(result['education'].unique())
    job_map = {value: index for index, value in enumerate(reference['job_categories'])}
    education_map = {
        value: index for index, value in enumerate(reference['education_categories'])
    }
    result['job_enc'] = result['job_category'].map(job_map).fillna(-1).astype(int)
    result['edu_enc'] = result['education'].map(education_map).fillna(-1).astype(int)
    return result


feature_reference = fit_feature_reference(train_raw)
train_features = transform_applicants(train_raw, feature_reference, training_rows=True)
validation_features = transform_applicants(validation_raw, feature_reference)
test_features = transform_applicants(test_raw, feature_reference)

feature_columns = [
    'age', 'work_experience', 'declared_income', 'job_enc', 'edu_enc',
    'address_change_count', 'num_applications', 'device_share_log',
    'ip_share_log', 'phone_share_log', 'age_exp_gap', 'income_per_exp',
    'income_per_age', 'rule_score', 'identity_coherence_score'
]
X_train = train_features[feature_columns]
y_train = train_features['label'].astype(int)
X_validation = validation_features[feature_columns]
y_validation = validation_features['label'].astype(int)
X_test = test_features[feature_columns]
y_test = test_features['label'].astype(int)

corrected_xgb_model = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=(y_train == 0).sum() / max((y_train == 1).sum(), 1),
    random_state=EVAL_SEED,
    eval_metric='auc',
    n_jobs=-1,
)
corrected_xgb_model.fit(X_train, y_train)

validation_probabilities = corrected_xgb_model.predict_proba(X_validation)[:, 1]
precision_values, recall_values, candidate_thresholds = precision_recall_curve(
    y_validation, validation_probabilities
)
validation_f1 = (
    2 * precision_values[:-1] * recall_values[:-1] /
    (precision_values[:-1] + recall_values[:-1] + 1e-12)
)
selected_threshold = float(candidate_thresholds[np.argmax(validation_f1)])

# The test set is evaluated only after model fitting and threshold selection.
test_probabilities = corrected_xgb_model.predict_proba(X_test)[:, 1]
test_predictions = (test_probabilities >= selected_threshold).astype(int)
test_matrix = confusion_matrix(y_test, test_predictions, labels=[0, 1])

print('Leakage-aware holdout evaluation')
print('Split sizes:', {'train': len(y_train), 'validation': len(y_validation), 'test': len(y_test)})
print('Train labels (genuine, fraud):', np.bincount(y_train, minlength=2).tolist())
print('Validation labels (genuine, fraud):', np.bincount(y_validation, minlength=2).tolist())
print('Test labels (genuine, fraud):', np.bincount(y_test, minlength=2).tolist())
print('Validation-selected threshold:', round(selected_threshold, 6))
print('Test ROC-AUC:', round(roc_auc_score(y_test, test_probabilities), 4))
print('Test PR-AUC:', round(average_precision_score(y_test, test_probabilities), 4))
print('Test precision:', round(precision_score(y_test, test_predictions, zero_division=0), 4))
print('Test recall:', round(recall_score(y_test, test_predictions, zero_division=0), 4))
print('Test F1:', round(f1_score(y_test, test_predictions, zero_division=0), 4))
print('Test confusion matrix [ [TN, FP], [FN, TP] ]:')
print(test_matrix)
```
