```python
from pathlib import Path

# Required external inputs for Module C. No data is downloaded automatically.
required_inputs = [
    Path('IBM Transaction Dataset/HI-Small_Trans.csv'),
    Path('BAF NeurIPS Dataset/Base.csv'),
]
missing_inputs = [path for path in required_inputs if not path.is_file()]

if missing_inputs:
    missing_list = '\n'.join(f'- {path.as_posix()}' for path in missing_inputs)
    raise FileNotFoundError(
        'Module C requires these external dataset files in the repository root:\n'
        f'{missing_list}\n'
        'Place the authorized dataset files at those paths, then rerun this cell.'
    )

for dataset_dir in ('IBM Transaction Dataset', 'BAF NeurIPS Dataset'):
    print(f'=== {dataset_dir} ===')
    for path in sorted(Path(dataset_dir).iterdir()):
        if path.is_file():
            print(path.name)

# Any saved output displayed below is historical until this cell is rerun.
```
